```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
image = cv2.imread('noisy_image_sample.jpg')
print("Image loaded successfully!")
print(f"Image shape: {image.shape}")
cv2_imshow(image)
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(image)
```

```python
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
image2 = clahe.apply(image)
cv2_imshow(image2)
```

```python
image3 = cv2.GaussianBlur(image2, (3, 3), 0)
cv2_imshow(image3)
```

```python
image4 = cv2.fastNlMeansDenoising(image3, None, 10, 7, 21)
cv2_imshow(image4)
```

```python
image5 = cv2.medianBlur(image4, 3)
cv2_imshow(image5)
```

```python
image6 = cv2.adaptiveThreshold(image5, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)
cv2_imshow(image6)
```

```python
cv2.imwrite('cleaned_image.jpg', image5)

from google.colab import files
files.download('cleaned_image.jpg')
```
