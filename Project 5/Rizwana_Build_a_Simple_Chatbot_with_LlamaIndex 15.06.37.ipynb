```python
import os
from google.colab import userdata

GROQ_API_KEY = userdata.get('Groq')
os.environ["GROQ_API_KEY"] = GROQ_API_KEY

if not GROQ_API_KEY:
  print("Warning: Please enter your API Key to Colab Secrets under the name 'Groq'.")
  print("You can retrieve a free key via https://console.groq.com/keys")

else:
  print("Groq Key activated!")
```

```python
from llama_index.llms.groq import Groq
from llama_index.core.llms import ChatMessage
```

```python
llm = Groq(model="llama-3.1-8b-instant")
```

```python
def simple_chatbot():
  print("Simple Groq Chatbot")
  print("Type 'exit' to end the conversation")
  print("-" * 50)

  messages = []

  while True:
    user_input = input("\nYou:  ")

    if user_input.lower() in ["exit", "quit", "bye"]:
      print("\nChatbot: Goodbye! See you later.")
      break

    if not user_input.strip():
      print("(Please type a question or type 'exit' to quit.)")
      continue

    messages.append(ChatMessage(role="user", content=user_input))


    try:
      response = llm.chat(messages)
      print(f"\nChatbot: {response.message.content}")
      messages.append(ChatMessage(role="assistant", content=response.message.content))

    except Exception as e:
      print(f"\nError: {e}")
      print("Please try again or check your API key and your Colab Secrets!")
```

```python
if __name__ == "__main__":
  simple_chatbot()
```
