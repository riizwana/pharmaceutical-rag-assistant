```python
from llama_index.core import SimpleDirectoryReader

documents = SimpleDirectoryReader("sample_docs").load_data()
print(f"Loaded {len(documents)} documents.")
```

```python
for i, doc in enumerate(documents):
  print(f"\n--- Document {i+1}: {doc.metadata['file_name']} ---")
  print(f"Length: {len(doc.text)} characters")
  print(doc.text[:200] + "...")
```

```python
from llama_index.core.node_parser import SentenceSplitter

splitter_fixed = SentenceSplitter(chunk_size=300, chunk_overlap=0)
chunks_fixed = splitter_fixed.get_nodes_from_documents(documents)
print(f"Total Fixed-Length Chunks Created: {len(chunks_fixed)}")
```

```python
splitter_small_test = SentenceSplitter(chunk_size=300, chunk_overlap=50)
chunks_overlap = splitter_small_test.get_nodes_from_documents(documents)
print(f"Total Small Test Chunks Created: {len(chunks_overlap)}")

print(f"\nSample small chunk 1: {chunks_overlap[0].text}")
print(f"\nSample small chunk 2: {chunks_overlap[1].text}")
```

```python
from llama_index.core.node_parser import SemanticSplitterNodeParser
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

embed_model = HuggingFaceEmbedding(model_name="sentence-transformers/all-MiniLM-L6-v2")

semantic_splitter = SemanticSplitterNodeParser(embed_model=embed_model)
chunks_semantic = semantic_splitter.get_nodes_from_documents(documents)
print(f"Total Semantic Chunks Created: {len(chunks_semantic)}")
```

```python
print(f"\nSample semantic chunk 1:")
print(chunks_semantic[0].text)
print(f"\nSample semantic chunk 2:")
print(chunks_semantic[1].text)
```

```python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

embed_model = HuggingFaceEmbedding(model_name="sentence-transformers/all-MiniLM-L6-v2")

for chunk in chunks_overlap:
  chunk.embedding = embed_model.get_text_embedding(chunk.text)

print("Embeddings Generated Successfully!")
```
